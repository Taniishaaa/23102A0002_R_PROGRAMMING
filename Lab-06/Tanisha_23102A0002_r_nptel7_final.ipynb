## Lecture 27 — Sequences of Dates and Alphabets

`as.Date()` creates Date objects. `by` controls the date interval.


### First day of each year


```r
seq(as.Date("2010-01-01"), as.Date("2017-01-01"), by = "years")
```

### Six yearly dates


```r
seq(as.Date("2017-01-01"), by = "years", length = 6)
```

### Six daily dates


```r
seq(as.Date("2017-01-01"), by = "days", length = 6)
```

### Six monthly dates


```r
seq(as.Date("2017-01-01"), by = "months", length = 6)
```

### Date sequence using start and end dates


```r
startdate = as.Date("2016-1-1")
```

```r
enddate = as.Date("2017-1-1")
```

```r
out = seq(enddate, startdate, by = "-1 month")
```

```r
out
```

## Lecture 28 — Repeats

`rep()` repeats values. `times` repeats the whole object; `each` repeats each element.


### `rep()` help


```r
help("rep")
```

### Repeat a value 10 times


```r
rep(3.5, times=10)
```

### Repeat a vector twice


```r
rep(1:4, 2)
```

### `times` versus `each`


```r
x = 1:4
```

```r
x
```

```r
rep(x, times = 3)
```

```r
rep(x, each = 3)
```

### Combine `each` and `times`


```r
rep(1:4, each = 2)
```

```r
rep(1:4, each = 2, times = 3)
```

```r
rep(1:4, times = 3, each = 2)
```

### Different repetition counts


```r
rep(1:4, 2:5)
```

```r
ans = seq(from=2, to=8, by=2)
```

```r
ans
```

```r
rep(1:4, ans)
```

### Repeat a matrix

```r
x
```

```r
rep(x, 3)
```

### Repeat character values


```r
rep(c("a", "b", "c"), 2)
```

```r
rep(c("apple", "banana", "cake"), 2)
```

### Repeat to a specified length

```r
rep("apple", length=5)
```

## Lecture 29 — Sorting, Ordering and Mode

`sort()` returns sorted values. `order()` returns the positions that produce the sorted order.


```r
y = c(8,5,7,6)
```

```r
y
```

```r
sort(y)
```

```r
sort(y, decreasing = TRUE)
```

### Sorting with missing values
`na.last=TRUE` puts `NA` last, `FALSE` puts it first, and `NA` removes missing values.


```r
sort(c(8, NA, 5, 7), na.last=TRUE)
```

```r
sort(c(8, NA, 5, 7), na.last=FALSE)
```

```r
sort(c(8, NA, 5, 7), na.last=NA)
```

### `order()`


```r
y = c(9,8,5,7,6)
```

```r
y
```

```r
order(y)
```

```r
order(y, decreasing = TRUE)
```

### `mode()`

```r
mode(c(3,4,5,6,7,8))
```

```r
mode("India")
```

```r
mode(c("India", "CANADA"))
```

## Lecture 30 — Lists

A list can contain objects of different types, including vectors, matrices, data frames and other lists.


### Matrices used in the examples


```r
x1 = matrix(nrow=2, ncol=2, data=1:4, byrow=T)
```

```r
x2 = matrix(nrow=2, ncol=2, data=5:8, byrow=T)
```

```r
x1
```

```r
x2
```

```r
x1+x2
```

### Put character data into a numeric matrix
A matrix has one common data type, so the numeric values become character.


```r
x1[2,1] = "hello"
```

```r
x1
```

```r
x1 = matrix(nrow=2, ncol=2, data=1:4, byrow=T)
```

```r
x2 = matrix(nrow=2, ncol=2, data=5:8, byrow=T)
```

### Create a list containing the two matrices


```r
matlist = list(x1, x2)
```

```r
matlist
```

### Access list elements with `[ ]`
Single brackets return a sublist.


```r
matlist[1]
```

```r
matlist[2]
```

### List containing different object types
The elements are a character vector, a numeric vector and a matrix.


```r
z1 = list( c("water", "juice", "lemonade"), rep(1:4, each=2), matrix(data=5:8, nrow=2, ncol=2, byrow=T) )
```

```r
z1
```

### Access an element with `[[ ]]`
Double brackets extract the actual list element.


```r
z1[[1]]
```

### `[ ]` versus `[[ ]]`
The first form operates on a sublist. The second extracts the vector first and then selects its second value.


```r
z1[1][2]
```

```r
z1[[1]][2]
```

### Named list elements
Named elements can be accessed with either `[[ ]]` or `$`.


```r
students = list(Students=c("A", "B"), Marks=c(90, 85))
```

```r
students[["Students"]]
```

```r
students$Students
```
