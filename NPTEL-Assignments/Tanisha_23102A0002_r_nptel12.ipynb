## Programming Example 1


```r
rm(list = ls())
```

```r
x = c(10, 20, 30)
y = c(1, 2, 3)
```

```r
example1 = function(x, y) {
  n = length(x)
  x1 = 0; y1 = 0; z1 = 0
  for (i in 1:n) {
    x1[i] = x[i]^2
    y1[i] = y[i]^2
    z1[i] = (x[i] / y[i])^2
  }
  sum_square_x = sum(x1)
  sum_square_y = sum(y1)
  sum_square_z = sum(z1)
  g = sum_square_x / sum_square_y
  h = sum_square_z
  cat("The value of g and h are", g, "and", h, "respectively", "\n")
}
```

```r
example1(x, y)
```

```r
x = c(67, 87, 26, 85, 6, 45)
y = c(54, 64, 22, 94, 20, 88)
example1(x, y)
```

```r
g = function(x, y) sum(x^2) / sum(y^2)
h = function(x, y) sum(x / y)^2
g(x, y)
h(x, y)
```

## Programming Example 3


```r
f = function(x) {
  if (x > 0) {
    exp((x + log(1 + x^3)) / x^2)
  } else if (x == 0) {
    10
  } else {
    (2 + x^3) / x
  }
}
```

```r
f(123)
f(-123)
f(0)
f(8)
f(-4)
```

```r
h = function() {
  x = seq(-1, 5, by = 0.2)
  y = numeric(length(x))
  for (i in 1:length(x)) {
    y[i] = f(x[i])
  }
  plot(x, y, type = "l")
}
```

```r
h()
```
