## Conditional Execution: if


```r
x = 5
if (x > 4) x * 3
```

```r
x = 3
if (x > 4) x * 3
```

```r
x = 5
if (x > 4) {x * 3}
```

```r
x = 3
if (x > 4) {x * 3}
```

Example: print when the value is more than 3.

```r
x = 6
if (x > 3) {
  print("The value is more than 3")
}
```

Example where the condition is false.

```r
x = 2
if (x > 3) {
  print("The value is more than 3")
}
```

## Conditional Execution: if-else


```r
x = 5
if (x == 3) { x = x - 1 } else { x = 2*x }
x
```

```r
x = 3
if (x == 3) { x = x - 1 } else { x = 2*x }
x
```

Example with x = 6.

```r
x = 6
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

Example with x = 2.

```r
x = 2
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

## Nested if-else if


```r
x = 5
if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2*x
}
x
```

```r
x = 2
if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2*x
}
x
```

```r
x = 3
if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2*x
}
x
```

## Conditional Execution: ifelse


```r
x = 1:10
x
ifelse(x < 6, x^2, x + 1)
```

```r
x = c(7, 9, 8, 4)
ifelse(x %% 2 == 0, "even number", "odd number")
```

## Conditional Execution: which


```r
x = c(10, 15, 8, 14, 6, 12)
x
which(x == 14)
```

```r
x = c(10, 15, 8, 14, 6, 12)
which(x != 12)
```

```r
x = c(10, 15, 8, 14, 6, 12)
which(x > 10)
```

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x
which.min(x)
```

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
which.max(x)
```

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x
which(x %% 2 == 1)
```

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
which(x %% 2 == 1, arr.ind = TRUE)
```
